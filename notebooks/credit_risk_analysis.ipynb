```python
history_features = (
    previous_loans
    .groupby("customerid")
    .agg(
        historical_loan_count=("systemloanid", "size"),
        first_historical_loan_date=("creationdate", "min"),
        last_historical_approval_date=("approveddate", "max")
    )
    .reset_index()
)
```

```python
history_features = history_features.merge(
    performance[["customerid", "creationdate"]],
    on="customerid",
    how="left",
    validate="one_to_one"
)
```

```python
history_features["history_duration_days"] = (
    history_features["creationdate"]
    - history_features["first_historical_loan_date"]
).dt.total_seconds() / (24 * 60 * 60)

history_features["days_since_last_loan"] = (
    history_features["creationdate"]
    - history_features["last_historical_approval_date"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
repayment_history = previous_loans.merge(
    performance[["customerid", "creationdate"]],
    on="customerid",
    how="left",
    validate="many_to_one",
    suffixes=("_historical", "_current")
)
```

```python
repayment_history["repayment_observed"] = (
    repayment_history["firstrepaiddate"]
    < repayment_history["creationdate_current"]
)
```

```python
repayment_history.head()
```

```python
repayment_history["repayment_delay_days"] = (
    repayment_history["firstrepaiddate"]
    - repayment_history["firstduedate"]
).dt.total_seconds() / (24 * 60 * 60)
```

```python
repayment_history.head()
```

```python
repayment_history["repayment_status"] = "not_observed"

repayment_history.loc[
    repayment_history["repayment_observed"]
    & (repayment_history["repayment_delay_days"] < 0),
    "repayment_status"
] = "early"

repayment_history.loc[
    repayment_history["repayment_observed"]
    & (repayment_history["repayment_delay_days"] == 0),
    "repayment_status"
] = "on_time"

repayment_history.loc[
    repayment_history["repayment_observed"]
    & (repayment_history["repayment_delay_days"] > 0),
    "repayment_status"
] = "late"
```

```python
repayment_history.head()
```

```python
repayment_features = (
    repayment_history
    .groupby("customerid")
    .agg(
        observed_repayment_count=(
            "repayment_observed",
            "sum"
        ),
        observed_early_repayment_count=(
            "repayment_status",
            lambda x: (x == "early").sum()
        ),
        observed_on_time_repayment_count=(
            "repayment_status",
            lambda x: (x == "on_time").sum()
        ),
        observed_late_repayment_count=(
            "repayment_status",
            lambda x: (x == "late").sum()
        ),
        avg_observed_repayment_delay_days=(
            "repayment_delay_days",
            lambda x: x[
                repayment_history.loc[x.index, "repayment_observed"]
            ].mean()
        ),
        avg_observed_late_days=(
            "repayment_delay_days",
            lambda x: x[
                repayment_history.loc[x.index, "repayment_status"] == "late"
            ].mean()
        )
    )
    .reset_index()
)
```

```python
repayment_features.head()
```

```python
repayment_features.shape
```

```python
repayment_features["observed_late_repayment_rate"] = (
    repayment_features["observed_late_repayment_count"]
    / repayment_features["observed_repayment_count"]
)

repayment_features["observed_early_repayment_rate"] = (
    repayment_features["observed_early_repayment_count"]
    / repayment_features["observed_repayment_count"]
)

repayment_features["observed_on_time_repayment_rate"] = (
    repayment_features["observed_on_time_repayment_count"]
    / repayment_features["observed_repayment_count"]
)
```

```python
outlier_cols = [
    "loanamount",
    "loan_cost",
    "loan_cost_rate",
    "historical_loan_count",
    "history_duration_days",
    "days_since_last_loan",
    "avg_historical_loan_amount",
    "total_historical_loan_amount",
    "avg_observed_repayment_delay_days",
    "avg_observed_late_days"
]

outlier_results = []

for col in outlier_cols:
    q1 = modeling_data[col].quantile(0.25)
    q3 = modeling_data[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outlier_mask = (
        (modeling_data[col] < lower)
        | (modeling_data[col] > upper)
    )

    outlier_results.append({
        "feature": col,
        "Q1": q1,
        "Q3": q3,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": outlier_mask.sum(),
        "outlier_pct": outlier_mask.mean() * 100
    })

outlier_summary = pd.DataFrame(outlier_results)

print("\nOUTLIER SUMMARY")
print(outlier_summary)
```

```python
# RISK-ORIENTED EVALUATION FUNCTION


from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

def evaluate_bad_risk(
    y_true,
    bad_probability,
    model_name,
    threshold=0.50
):
    """
    Evaluate a model with Bad treated as the positive event.

    Official project encoding:
        Good = 1
        Bad  = 0

    Therefore:
        bad_probability = P(Bad)
    """

    # Convert probability into risk classification
    bad_prediction = (bad_probability >= threshold).astype(int)

    # Convert to actual class labels:
    # Bad = 0, Good = 1
    y_pred = np.where(bad_prediction == 1, 0, 1)

    # Convert truth into Bad=1 for risk metrics
    y_true_bad = (y_true == 0).astype(int)

    # Evaluate against Bad-as-positive representation
    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true_bad,
        bad_prediction,
        zero_division=0
    )

    recall = recall_score(
        y_true_bad,
        bad_prediction,
        zero_division=0
    )

    f1 = f1_score(
        y_true_bad,
        bad_prediction,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_true_bad,
        bad_probability
    )

    pr_auc = average_precision_score(
        y_true_bad,
        bad_probability
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    print(f"\n{'=' * 60}")
    print(model_name)
    print(f"{'=' * 60}")

    print(f"Threshold: {threshold:.2f}")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1:        {f1:.4f}")
    print(f"ROC-AUC:   {roc_auc:.4f}")
    print(f"PR-AUC:    {pr_auc:.4f}")

    print("\nConfusion Matrix")
    print("Rows = Actual [Bad, Good]")
    print("Cols = Predicted [Bad, Good]")
    print(cm)

    return {
        "model": model_name,
        "threshold": threshold,
        "accuracy": accuracy,
        "precision_bad": precision,
        "recall_bad": recall,
        "f1_bad": f1,
        "roc_auc": roc_auc,
        "pr_auc": pr_auc,
        "confusion_matrix": cm
    }
```

```python
# THRESHOLD ANALYSIS


def threshold_analysis(
    y_true,
    bad_probability,
    model_name,
    thresholds=None
):
    if thresholds is None:
        thresholds = np.arange(0.10, 0.61, 0.05)

    y_true_bad = (y_true == 0).astype(int)

    rows = []

    for threshold in thresholds:

        predicted_bad = (
            bad_probability >= threshold
        ).astype(int)

        precision = precision_score(
            y_true_bad,
            predicted_bad,
            zero_division=0
        )

        recall = recall_score(
            y_true_bad,
            predicted_bad,
            zero_division=0
        )

        f1 = f1_score(
            y_true_bad,
            predicted_bad,
            zero_division=0
        )

        flagged_rate = predicted_bad.mean()

        rows.append({
            "Model": model_name,
            "Threshold": threshold,
            "Precision_Bad": precision,
            "Recall_Bad": recall,
            "F1_Bad": f1,
            "Flagged_as_Bad_%": flagged_rate * 100
        })

    return pd.DataFrame(rows)
```

```python
# BUSINESS COST SENSITIVITY


def threshold_cost_analysis(
    y_true,
    bad_probability,
    model_name,
    thresholds=None,
    fn_costs=(1, 2, 5, 10)
):
    """
    Sensitivity analysis for different hypothetical
    False Negative : False Positive cost ratios.

    FN = actual Bad predicted Good
    FP = actual Good predicted Bad
    """

    if thresholds is None:
        thresholds = np.arange(0.10, 0.61, 0.01)

    y_true_bad = (y_true == 0).astype(int)

    rows = []

    for threshold in thresholds:

        predicted_bad = (
            bad_probability >= threshold
        ).astype(int)

        fn = (
            (y_true_bad == 1) &
            (predicted_bad == 0)
        ).sum()

        fp = (
            (y_true_bad == 0) &
            (predicted_bad == 1)
        ).sum()

        tn = (
            (y_true_bad == 0) &
            (predicted_bad == 0)
        ).sum()

        tp = (
            (y_true_bad == 1) &
            (predicted_bad == 1)
        ).sum()

        base_row = {
            "Model": model_name,
            "Threshold": threshold,
            "TP": tp,
            "FN": fn,
            "FP": fp,
            "TN": tn
        }

        for fn_cost in fn_costs:
            fp_cost = 1

            expected_cost = (
                fn * fn_cost +
                fp * fp_cost
            )

            base_row[f"Cost_FN{fn_cost}_FP1"] = expected_cost

        rows.append(base_row)

    return pd.DataFrame(rows)
```

```python
cost_logistic = threshold_cost_analysis(
    y_validation,
    balanced_logistic_bad_probability,
    "Balanced Logistic Regression"
)

cost_forest = threshold_cost_analysis(
    y_validation,
    balanced_forest_bad_probability,
    "Balanced Random Forest"
)

cost_analysis = pd.concat(
    [cost_logistic, cost_forest],
    ignore_index=True
)

cost_analysis.round(3)
```

```python
# TIME-AWARE BAD-RISK SCORER


from sklearn.metrics import (
    precision_recall_curve,
    auc
)

def bad_pr_auc_scorer(estimator, X, y):
    """
    Exact PR-AUC scorer with Bad treated as the
    positive event.

    Official project encoding:
        Good = 1
        Bad  = 0
    """

    bad_probability = estimator.predict_proba(X)[:, 0]

    y_bad = (y == 0).astype(int)

    precision, recall, _ = precision_recall_curve(
        y_bad,
        bad_probability
    )

    return auc(recall, precision)
```

```python
# SUBGROUP PERFORMANCE AUDIT


def subgroup_audit(
    data,
    group_column,
    min_group_size=30
):

    rows = []

    for group, group_data in data.groupby(
        group_column,
        dropna=False
    ):

        n = len(group_data)

        if n < min_group_size:
            continue

        actual_bad = group_data["actual_bad"]
        predicted_bad = group_data["predicted_bad"]

        tp = (
            (actual_bad == 1) &
            (predicted_bad == 1)
        ).sum()

        fn = (
            (actual_bad == 1) &
            (predicted_bad == 0)
        ).sum()

        fp = (
            (actual_bad == 0) &
            (predicted_bad == 1)
        ).sum()

        tn = (
            (actual_bad == 0) &
            (predicted_bad == 0)
        ).sum()

        bad_count = actual_bad.sum()

        bad_rate = (
            bad_count / n
            if n > 0
            else np.nan
        )

        recall_bad = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else np.nan
        )

        false_positive_rate = (
            fp / (fp + tn)
            if (fp + tn) > 0
            else np.nan
        )

        precision_bad = (
            tp / (tp + fp)
            if (tp + fp) > 0
            else np.nan
        )

        rows.append({
            "Group": group,
            "N": n,
            "Bad_Rate": bad_rate,
            "Mean_Predicted_Bad": group_data[
                "bad_probability"
            ].mean(),
            "Recall_Bad": recall_bad,
            "False_Positive_Rate": false_positive_rate,
            "Precision_Bad": precision_bad,
            "Predicted_Bad_Rate": predicted_bad.mean()
        })

    return (
        pd.DataFrame(rows)
        .sort_values("N", ascending=False)
        .reset_index(drop=True)
    )
```

```python
# DETAILED SUBGROUP ERROR AUDIT


def detailed_subgroup_audit(
    data,
    group_column,
    threshold=0.23,
    min_group_size=30
):
    
    rows = []

    for group, group_data in data.groupby(
        group_column,
        dropna=False
    ):

        n = len(group_data)

        if n < min_group_size:
            continue

        actual_bad = group_data["actual_bad"].astype(int)
        predicted_bad = group_data["predicted_bad"].astype(int)

        tp = (
            (actual_bad == 1) &
            (predicted_bad == 1)
        ).sum()

        fn = (
            (actual_bad == 1) &
            (predicted_bad == 0)
        ).sum()

        fp = (
            (actual_bad == 0) &
            (predicted_bad == 1)
        ).sum()

        tn = (
            (actual_bad == 0) &
            (predicted_bad == 0)
        ).sum()

        actual_bad_count = tp + fn
        actual_good_count = fp + tn

        rows.append({
            "Group": group,
            "N": n,
            "Actual_Bad": actual_bad_count,
            "Actual_Good": actual_good_count,
            "TP": tp,
            "FN": fn,
            "FP": fp,
            "TN": tn,
            "Bad_Rate": actual_bad.mean(),
            "Recall_Bad": (
                tp / actual_bad_count
                if actual_bad_count > 0
                else np.nan
            ),
            "False_Positive_Rate": (
                fp / actual_good_count
                if actual_good_count > 0
                else np.nan
            ),
            "Precision_Bad": (
                tp / (tp + fp)
                if (tp + fp) > 0
                else np.nan
            ),
            "Predicted_Bad_Rate": predicted_bad.mean(),
            "Mean_Predicted_Bad": (
                group_data["bad_probability"].mean()
            )
        })

    return (
        pd.DataFrame(rows)
        .sort_values("N", ascending=False)
        .reset_index(drop=True)
    )
```

```python
# LOCAL SENSITIVITY EXPLANATION

def explain_single_loan(
    model,
    X_data,
    row_index,
    numeric_features,
    categorical_features,
    top_n=10
):
    """
    Local what-if sensitivity analysis.

    For each feature:
        replace the feature with a reference value,
        re-score the observation,
        measure the change in estimated Bad risk.

    This is a sensitivity analysis, not a causal explanation.
    """

    original = X_data.iloc[row_index].copy()

    baseline_probability = model.predict_proba(
        pd.DataFrame(
            [original],
            columns=X_data.columns
        )
    )[0, 0]

    sensitivities = []

    for feature in numeric_features:

        perturbed = original.copy()

        reference_value = (
            X_train[feature].median()
        )

        perturbed[feature] = reference_value

        perturbed_probability = model.predict_proba(
            pd.DataFrame(
                [perturbed],
                columns=X_data.columns
            )
        )[0, 0]

        change = (
            perturbed_probability -
            baseline_probability
        )

        sensitivities.append({
            "Feature": feature,
            "Original_Value": original[feature],
            "Reference_Value": reference_value,
            "Risk_Change_When_Reset": change
        })

    for feature in categorical_features:

        perturbed = original.copy()

        reference_value = (
            X_train[feature]
            .mode(dropna=True)
            .iloc[0]
        )

        perturbed[feature] = reference_value

        perturbed_probability = model.predict_proba(
            pd.DataFrame(
                [perturbed],
                columns=X_data.columns
            )
        )[0, 0]

        change = (
            perturbed_probability -
            baseline_probability
        )

        sensitivities.append({
            "Feature": feature,
            "Original_Value": original[feature],
            "Reference_Value": reference_value,
            "Risk_Change_When_Reset": change
        })

    explanation = pd.DataFrame(sensitivities)

    explanation["Absolute_Risk_Change"] = (
        explanation["Risk_Change_When_Reset"].abs()
    )

    explanation = (
        explanation
        .sort_values(
            "Absolute_Risk_Change",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )

    return baseline_probability, explanation
```
